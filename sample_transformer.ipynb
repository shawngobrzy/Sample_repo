```python
import torch
import torch.nn as nn
from torch.nn import functional as F

#Hyperparameters 
batch_size = 16        #How many sequences we process in parallel
block_size = 32        #Maximum context length the model looks back on
max_iters = 5000       #Number of training iterations
eval_interval = 100    #How often to compute train/val loss
learning_rate = 1e-3
device = 'cpu'
eval_iters = 200
n_embd = 64            #Embedding dimension
n_head = 4             #Number of attention heads
n_layer = 4            #Number of transformer blocks
dropout = 0.0

torch.manual_seed(1337)
```

```python
#Load up the data
with open('testdrive.txt', 'r', encoding='utf-8') as f:
    text = f.read()

#Find all unique characters in the dataset
characters = sorted(list(set(text)))
unique_characters = len(characters)

#Character to Integer Map transformation
string_to_int = { ch:i for i,ch in enumerate(characters) }
int_to_string = { i:ch for i,ch in enumerate(characters) }
encode = lambda s: [string_to_int[c] for c in s]          #String to list[int]
decode = lambda l: ''.join([int_to_string[i] for i in l]) #List[int] to string

#Convert the entire dataset into integers
data = torch.tensor(encode(text), dtype=torch.long)

#Make a 90/10 split on training and testing data
n = int(0.9*len(data))
train_data = data[:n]
test_data = data[n:]
```

```python
#Batching function
def get_batch(split):
    """
    Returns:
        x: (batch_size, block_size)  input sequences
        y: (batch_size, block_size)  the same sequences shifted by 1 token

    y is the 'targets': the correct next character for each position in x.
    """
    data = train_data if split == 'train' else test_data

    #Create a random starting position
    ix = torch.randint(len(data) - block_size, (batch_size,)) ### MAKE SURE TO PUT THE COMMA AFTER BATCH SIZE BECAUSE IT IS INDEED A TUPLE ###

    #X is training data or the current text (Input sequence)
    #Y is the target or the text with the desired next-token label (Same input with the next target label)
    x = torch.stack([data[i:i+block_size] for i in ix])
    y = torch.stack([data[i+1:i+block_size+1] for i in ix])

    return x.to(device), y.to(device)
```

```python
#Calculate average loss on training and testing sets: To monitor training without affecting gradients
@torch.no_grad()
def estimate_loss():
    
    out = {}
    model.eval()

    for split in ['train', 'val']:
        losses = torch.zeros(eval_iters)
        for k in range(eval_iters):
            X, Y = get_batch(split)
            logits, loss = model(X, Y)
            losses[k] = loss.item()
        out[split] = losses.mean()

    model.train()
    return out
```

```python
#Self-Attention Head 
class Head(nn.Module):
    """ One head of self-attention """

    def __init__(self, head_size):
        super().__init__()
        #Linear projection of key, query, and value
        self.key   = nn.Linear(n_embd, head_size, bias=False)
        self.query = nn.Linear(n_embd, head_size, bias=False)
        self.value = nn.Linear(n_embd, head_size, bias=False)

        #Mask ensures that future tokens are not seen (No cheating)
        self.register_buffer('tril', torch.tril(torch.ones(block_size, block_size)))
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        B, T, C = x.shape

        #Compute the key and query
        k = self.key(x)
        q = self.query(x)

        #Attention scores
        wei = q @ k.transpose(-2, -1) * (C ** -0.5)

        #Mask future positions (causal attention)
        wei = wei.masked_fill(self.tril[:T, :T] == 0, float('-inf'))

        #Convert scores to probabilities
        wei = F.softmax(wei, dim=-1)
        wei = self.dropout(wei)

        #Weighted sum of values
        v = self.value(x)
        out = wei @ v
        return out
```

```python
#Multihead attention
class MultiHeadAttention(nn.Module):
    """ Multiple self-attention heads in parallel """

    def __init__(self, num_heads, head_size):
        super().__init__()
        #Create heads
        self.heads = nn.ModuleList([Head(head_size) for _ in range(num_heads)])
        #Projection to return to full embedding size
        self.proj = nn.Linear(n_embd, n_embd)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        #Concatenate each head's output
        out = torch.cat([h(x) for h in self.heads], dim=-1)
        #Final projection
        out = self.dropout(self.proj(out))
        return out
```

```python
#Feed Forward Network
class FeedFoward(nn.Module):
    """ Simple MLP used inside each transformer block """

    def __init__(self, n_embd):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_embd, 4 * n_embd),
            nn.ReLU(),
            nn.Linear(4 * n_embd, n_embd),
            nn.Dropout(dropout),
        )

    def forward(self, x):
        return self.net(x)
```

```python
#Transformer Block
class Block(nn.Module):
    """ Transformer block = attention + feed-forward """

    def __init__(self, n_embd, n_head):
        super().__init__()
        head_size = n_embd // n_head
        self.sa   = MultiHeadAttention(n_head, head_size)
        self.ffwd = FeedFoward(n_embd)
        self.ln1  = nn.LayerNorm(n_embd)
        self.ln2  = nn.LayerNorm(n_embd)

    def forward(self, x):
        #Residual connections: x + something
        x = x + self.sa(self.ln1(x))
        x = x + self.ffwd(self.ln2(x))
        return x
```

```python
#Language Model
class BigramLanguageModel(nn.Module):
    """
    This is a minimal GPT-like Transformer LM.
    It predicts the next character given previous characters.
    """

    def __init__(self):
        super().__init__()

        #Token embeddings: look up a vector for each character id
        self.token_embedding_table = nn.Embedding(unique_characters, n_embd)

        #Positional embeddings: tell the model *where* each token is in the sequence
        self.position_embedding_table = nn.Embedding(block_size, n_embd)

        #Stack several transformer blocks
        self.blocks = nn.Sequential(*[Block(n_embd, n_head) for _ in range(n_layer)])

        #Final normalization
        self.ln_f = nn.LayerNorm(n_embd)

        #Projection to vocabulary size or number of unique characters (logits for next character)
        self.lm_head = nn.Linear(n_embd, unique_characters)

    def forward(self, idx, targets=None):
        """
        idx:     (B, T) current input tokens
        targets: (B, T) the correct next tokens

        targets is used ONLY during training for loss computation.
        """

        B, T = idx.shape

        #Embed tokens and positions
        tok_emb = self.token_embedding_table(idx)                #(B, T, C) [Batch size, Sequence length, Number of classes {unique characters}]
        pos_emb = self.position_embedding_table(torch.arange(T, device=device)) #(T, C)

        #Combine the tokens and positions
        x = tok_emb + pos_emb

        #Pass through transformer blocks
        x = self.blocks(x)
        x = self.ln_f(x)

        #Final prediction for each token in the sequence
        logits = self.lm_head(x)                                #(B, T, unique_characters)

        if targets is None:
            #When generating, we don't compute loss
            return logits, None
        else:
            #Flatten both tensors so cross_entropy works
            B, T, C = logits.shape
            logits  = logits.view(B*T, C)
            targets = targets.view(B*T)

            #Cross-entropy compares predicted logits vs correct next token
            loss = F.cross_entropy(logits, targets)
            return logits, loss

    def generate(self, idx, max_new_tokens):
        """
        Generate new text by repeatedly predicting the next token.
        idx: starting sequence of token ids
        """
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -block_size:]       #Keep only recent context
            logits, _ = self(idx_cond)            #Forward pass
            logits = logits[:, -1, :]             #Take last timestep
            probs  = F.softmax(logits, dim=-1)    #Convert to probabilities
            idx_next = torch.multinomial(probs, 1) #Sample next token
            idx = torch.cat((idx, idx_next), dim=1)
        return idx
```

```python
#Big Language Model
model = BigramLanguageModel()
m = model.to(device)
```

```python
#Print parameter count
print(sum(p.numel() for p in m.parameters())/1e6, "M parameters")
```

```python
#Optimize PyTorch
optimizer = torch.optim.AdamW(model.parameters(), lr=learning_rate)
```

```python
#Training Loop
for iter in range(max_iters):

    #Regularly check train/val loss
    if iter % eval_interval == 0 or iter == max_iters - 1:
        losses = estimate_loss()
        print(f"step {iter}: train loss {losses['train']:.4f}, val loss {losses['val']:.4f}")

    #Get batch
    xb, yb = get_batch('train')

    #Forward pass
    logits, loss = model(xb, yb)

    #Update
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()
```
